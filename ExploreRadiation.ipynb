```python
import requests

from bs4 import BeautifulSoup


RADIATION_URL = (
    "https://bnra.bg/bg/byuletin-za-gama-fona/"
)

radiation_response = requests.get(
    RADIATION_URL,
    timeout=30,
)
#Check
print(radiation_response.status_code)
```

```python
# Convert
soup = BeautifulSoup(
    radiation_response.text,
    "html.parser",
)


varna_row = None
updated_row = None


for row in soup.find_all("tr"):
    cells = [
        cell.get_text(" ", strip=True)
        for cell in row.find_all(["th", "td"])
    ]

    if not cells:
        continue

    if cells[0].casefold() == "варна":
        varna_row = cells

        varna_table = row.find_parent("table")

        for table_row in varna_table.find_all("tr"):
            table_cells = [
                cell.get_text(" ", strip=True)
                for cell in table_row.find_all(["th", "td"])
            ]

            if (
                table_cells
                and "Актуални към" in table_cells[0]
            ):
                updated_row = table_cells
                break

        break


print("Varna row:", varna_row)
print("Updated row:", updated_row)
```
